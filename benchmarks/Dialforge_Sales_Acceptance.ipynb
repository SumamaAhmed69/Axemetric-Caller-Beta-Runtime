# Dialforge Sales Acceptance

This is the **consultative-sales acceptance test** for the upgraded Dialforge sales brain. It tests multi-turn cold calls, discovery, qualification, value framing, objection handling, direct pricing, closing, action integrity, DNC/follow-up/booking tool routing, and warmed local voice latency.

It compares **`qwen3:1.7b`**, **`qwen3:4b-instruct`**, and **`qwen3:4b-instruct-2507-q8_0`** and selects the model that behaves like the best salesperson, not simply the largest model.

### Run it
1. Choose **Runtime → Change runtime type → T4 GPU**.
2. Choose **Runtime → Run all**.
3. Leave the tab open until the report appears.

The benchmark uses a fictional, fully specified campaign dossier so it can distinguish strong selling from hallucination. No real calls or SIP credentials are used.

```python
# ONE-CLICK DIALFORGE SALES ACCEPTANCE v1
import base64, json, os, pathlib, subprocess, sys, time, urllib.request
from IPython.display import HTML, FileLink, display
OWNER='SumamaAhmed69'; REPO='Axemetric-Caller-Beta-Runtime'
BASE_BOOTSTRAP_PATH='benchmarks/dialforge_final_colab_bootstrap_v3.py'
BOOTSTRAP=pathlib.Path('/content/dialforge_sales_colab_bootstrap.py')
REPORT=pathlib.Path('/content/dialforge-sales-acceptance/dialforge-sales-acceptance.html')
JSON_REPORT=pathlib.Path('/content/dialforge-sales-acceptance/dialforge-sales-acceptance.json')
ZIP_REPORT=pathlib.Path('/content/dialforge-sales-acceptance-results.zip')
def api_json(url):
    req=urllib.request.Request(url,headers={'User-Agent':'Dialforge-Sales-Colab','Accept':'application/vnd.github+json'})
    with urllib.request.urlopen(req,timeout=60) as response: return json.loads(response.read().decode('utf-8'))
print('Resolving one immutable Dialforge source revision...')
source_sha=api_json(f'https://api.github.com/repos/{OWNER}/{REPO}/commits/main?x={time.time_ns()}')['sha']
print('Pinned source:',source_sha)
item=api_json(f'https://api.github.com/repos/{OWNER}/{REPO}/contents/{BASE_BOOTSTRAP_PATH}?ref={source_sha}&x={time.time_ns()}')
text=base64.b64decode(item['content']).decode('utf-8')
replacements={
    'RUNNER = ROOT / \"dialforge_final_acceptance_v3.py\"':'RUNNER = ROOT / \"dialforge_sales_acceptance.py\"',
    'REPORT_DIR = ROOT / \"dialforge-final-acceptance-v3\"':'REPORT_DIR = ROOT / \"dialforge-sales-acceptance\"',
    'REPORT_HTML = REPORT_DIR / \"dialforge-final-acceptance-v3.html\"':'REPORT_HTML = REPORT_DIR / \"dialforge-sales-acceptance.html\"',
    'REPORT_JSON = REPORT_DIR / \"dialforge-final-acceptance-v3.json\"':'REPORT_JSON = REPORT_DIR / \"dialforge-sales-acceptance.json\"',
    'REPORT_ZIP = ROOT / \"dialforge-final-acceptance-v3-results.zip\"':'REPORT_ZIP = ROOT / \"dialforge-sales-acceptance-results.zip\"',
    'RUNNER_PATH = \"benchmarks/dialforge_final_acceptance_v3.py\"':'RUNNER_PATH = \"benchmarks/dialforge_sales_acceptance.py\"',
    '=== DIALFORGE FINAL COLAB ACCEPTANCE BOOTSTRAP v3 ===':'=== DIALFORGE SALES COLAB ACCEPTANCE v1 ===',
    '=== DIALFORGE FINAL ACCEPTANCE v3 COMPLETE ===':'=== DIALFORGE SALES ACCEPTANCE COMPLETE ===',
    'Downloading tool-aware runner':'Downloading consultative-sales runner',
}
for old,new in replacements.items():
    if old not in text: raise RuntimeError(f'Expected bootstrap marker missing: {old}')
    text=text.replace(old,new)
compile(text,str(BOOTSTRAP),'exec'); BOOTSTRAP.write_text(text,encoding='utf-8')
print(f'Sales bootstrap verified: {len(text.encode())} bytes')
env=os.environ.copy(); env['DIALFORGE_SOURCE_SHA']=source_sha
print('\nStarting Dialforge Sales Acceptance...')
result=subprocess.run([sys.executable,str(BOOTSTRAP)],env=env)
if result.returncode!=0: raise RuntimeError('Sales acceptance stopped. The exact failing component is printed above.')
if not REPORT.exists() or not JSON_REPORT.exists() or not ZIP_REPORT.exists(): raise RuntimeError('Sales acceptance completed without all report files.')
print('\n=== DIALFORGE SALES ACCEPTANCE COMPLETE ===')
display(HTML(REPORT.read_text(encoding='utf-8')))
print('\nDownload/share these results:'); display(FileLink(str(ZIP_REPORT))); display(FileLink(str(JSON_REPORT)))
```

## What a great result means
A candidate earns `PHENOMENAL` only if it combines strong consultative sales judgment with correct tool routing and needs very little deterministic guard rescue. This benchmark intentionally rewards discovery, relevance, truthful objection handling and earned closes rather than aggressive pitching. A real Windows SIP/PSTN call remains the final telephony proof.
