# Lineborn v6 deployment — Drive-auth-safe version

Use **Runtime → Change runtime type → T4 GPU**, then **Runtime → Run all**.

This version tries Google Drive first. If Colab throws `credential propagation was unsuccessful`, it automatically switches to manual upload mode and asks for the 8 adapter parts one at a time.

The build still verifies the exact v6 archive SHA before merging or quantizing anything.


```python
import shutil, subprocess, os
from pathlib import Path

repo = Path('/content/lineborn-runtime')
if repo.exists():
    shutil.rmtree(repo)
subprocess.run([
    'git', 'clone', '--depth', '1', '--branch', 'lineborn-v6-deployment',
    'https://github.com/SumamaAhmed69/Axemetric-Caller-Beta-Runtime.git',
    str(repo),
], check=True)
os.chdir(repo)
print('Repository ready:', repo)
```

```python
%run training/lineborn_v6_colab_runner.py
```

## What happens if Drive auth is broken?

The runner will prompt for these files one by one:

`lineborn-sales-adapters-v6.zip.part1` through `part8`.

After the model build finishes, if Drive is still unavailable the outputs stay at `/content/lineborn-v6-export/`. Open the **Files** sidebar in Colab and download them from there.
