```python
# 1) Clone the validated deployment branch into temporary storage
import shutil, subprocess, os
from pathlib import Path

# Move to a stable directory before deleting a repo we may currently be inside.
os.chdir('/kaggle/working')
repo = Path('/tmp/lineborn-runtime')
if repo.exists():
    shutil.rmtree(repo)
subprocess.run([
    'git','clone','--depth','1','--branch','lineborn-v6-deployment',
    'https://github.com/SumamaAhmed69/Axemetric-Caller-Beta-Runtime.git',
    str(repo)
], check=True)
os.chdir(repo)
print('Repository ready:', repo)
```

```python
# 2) Build the trained Lineborn models\nimport runpy\nrunpy.run_path('training/lineborn_v6_kaggle_runner.py', run_name='__main__')\n
```

```python
# 3) Show the final files that Kaggle will persist
from pathlib import Path
out = Path('/kaggle/working/lineborn-v6-export')
print('Final Lineborn v6 artifacts:')
for p in sorted(out.iterdir()):
    if p.is_file():
        size = p.stat().st_size
        print(f'  {p.name}: {size/1024**3:.3f} GiB' if size > 1024**3 else f'  {p.name}: {size:,} bytes')
print('\nNow use Kaggle Save Version / Commit & Run so these outputs are persisted.')
```
